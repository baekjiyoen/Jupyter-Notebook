```python
import numpy as np
import pandas as pd
```

```python
txt={'파이썬 차트 파이썬 머신러닝',
    '차트 파이썬 R 차트',
    'R 분석 시각화'}
```

```python
txtList=[i.split() for i in txt]
txtList
```

```python
# 문서 전체에 등장하는 단어들 추출
nouns=sorted(list(set([j for i in txtList for j in i])))
nouns
```

```python
dtm=[]
for i in txtList:
    dtm.append([])
    for j in nouns:
        dtm[-1].append(i.count(j))
tf=pd.DataFrame(dtm,columns=nouns)
tf
```

```python
D=len(tf)
df=tf.astype('bool').sum()
idf=D/df
idf
```

```python
a=[3,4,0,5,1] # [1,1,0,1,1]
[bool(i) for i in a]
```

```python
#TF-IDF
tfidf=tf*idf
tfidf
```
