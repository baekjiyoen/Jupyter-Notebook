```python
import pandas as pd
from konlpy.tag import Okt
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df=pd.read_csv('./data/전처리_샘플뉴스.csv')
df
```

```python
text_res=[]
for i in df['text']:
    text_res.extend(i.split('\n'))
    break
text_res
```

```python
df['text'][0]
```

```python
# 문자별 명사 추출
okt=Okt()
nouns=[okt.nouns(i) for i in text_res]
```

```python
len(nouns)
```

```python
# stop_word
stop_word=['서울','단계']
last_nouns=[]
for i in nouns:
    last_nouns.append([j for j in i if j not in stop_word])

last_nouns
    
```

```python
# 빈줄 삭제
last_nouns=[i for i in last_nouns if len(i) !=0]
len(last_nouns)
last_nouns # 각 문서에서 명사 추출
```

```python
# 단어 집합
word_list=sorted(list(set([j for i in last_nouns for j in i])))
print(len(word_list))
word_list
```

```python
count=[]
for i in last_nouns:
    count.append([])
    for j in word_list:
        count[-1].append(i.count(j))
tf_df=pd.DataFrame(count,columns=word_list)
tf_df
```

```python
['가','가','나'].count('가')
```

```python
'가 가 나 '.count('가')
```

```python
# idf
D=len(tf_df)
# print(D)
df=tf_df.astype('bool').sum()
# df
idf=np.log(D/(df+1))
tfidf_df=tf_df*idf
tfidf_df
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cos_sim = cosine_similarity(tfidf_df,tfidf_df)
result_df = pd.DataFrame(cos_sim)
result_df
```
